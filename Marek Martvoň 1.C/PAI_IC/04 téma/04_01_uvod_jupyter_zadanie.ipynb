# Úvod do Jupyter notebookov vo VS Code

## Základné klávesové skratky

V Jupyter notebooku vo VS Code existujú dva režimy:
- **Edit mode** (v bunke bliká kurzor) - upravujete obsah bunky
- **Command mode** (bunka je označená, kurzor v nej nebliká) - pracujete s bunkami

### Prepínanie medzi režimami

- `Enter` - vstup do Edit mode
- `Esc` - vstup do Command mode

### Dôležité skratky v Command mode

- `A` - pridať bunku **nad** aktuálnu bunku (Above)
- `B` - pridať bunku **pod** aktuálnu bunku (Below)
- `M` - zmeniť bunku na Markdown
- `Y` - zmeniť bunku na Code
- `C` - kopírovať bunku (Copy)
- `V` - vložiť bunku (Paste)
- `X` - vystrihnúť bunku (Cut)
- `DD` - zmazať bunku (Delete - stlačte D dvakrát)
- `Z` - vrátiť späť posledný príkaz (Undo)

### Spustenie bunky

- `Shift + Enter` - spustiť bunku a prejsť na ďalšiu
- `Ctrl + Enter` - spustiť bunku a zostať v nej
